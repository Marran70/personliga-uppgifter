```python
from mtm026 import *
EI, L, x, Δ, q0 = sp.symbols("EI, L, x, Δ, q0")
```

definera diffekv

```python
q = q0 * x/L
w = sp.Function('w')(x)

ode = sp.Eq(sp.diff(EI*sp.diff(w, x, 2),  x, 2), q)
display(ode)
```

```python
w = sp.dsolve(ode, w).rhs
displayvar("W(X)", w)
```

# randvillkor

```python
wprim = w.diff(x)
wbis = w.diff(x, 2)
display(wbis)
M = -EI*wbis

r1 = sp.Eq(w.subs(x, 0), 0)
r2 = sp.Eq(wprim.subs(x, 0), 0)
r3 = sp.Eq(w.subs(x, L), 0)
r4 = sp.Eq(wprim.subs(x, L), 0)

display(r1, r2, r3, r4)
```

# lös ut varje konstant

```python
ekvationer = [r1, r2, r3, r4]
obekanta = sp.symbols("C1, C2, C3, C4")
sol = sp.solve(ekvationer, obekanta)
displayvar("sol", sol)
```

# beräkna M

```python
w = w.subs(sol)
displayvar("w(x)", w)

M = M.subs(sol)
displayvar("m(x)", M)
```
