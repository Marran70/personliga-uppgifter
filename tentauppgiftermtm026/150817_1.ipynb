```python
from mtm026 import *

E, I, P, M, L, =sp.symbols("E, I, P, M , L")

EI = E*L
K = sp.zeros(8, 8)
```

```python
ke1 = Ke_balk(EI=EI, L=L)
ke2 = Ke_balk(EI=EI, L = L/2)
assem(K, ke1, dofs=[1, 2, 3, 4])
assem(K, ke2, dofs=[3, 4, 5, 6])
assem(K, ke2, dofs=[5, 6, 7, 8])
display(K)
```

```python
a1, a2, a3, a4, a5, a6, a7, a8 = sp.symbols("a1:9")
r1, r2, r3, r4, r5, r6, r7, r8 = sp.symbols("r1:9")

a = sp.Matrix([0, 0, 0, a4, a5, a6, 0, a8])
r = sp.Matrix([r1, r2, r3, 0, 0, 0, r7, 0])

f = sp.zeros(8, 1)
f[4] = -P
f[7] = -M

ekv_sys = sp.Eq(K*a, f+r)
display(ekv_sys)
obekanta = [a4, a5, a6, a8, r1, r2, r3, r7]
sol = sp.solve(ekv_sys, obekanta)
display(sol[r2])
display(sol[a5])
```
