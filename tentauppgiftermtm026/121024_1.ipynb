```python
from mtm026 import *
L, E, I, q, q0, x = sp.symbols("L, E, I, q, q0, x")
EI = E*I
q = -q0*sp.sin(sp.pi*x/L)

w = sp.Function("w")(x)

diffekv = sp.Eq(sp.diff(EI*sp.diff(w, x, 2), x, 2), q)
display(diffekv)

sol = sp.dsolve(diffekv)
display(sol)
u = sol.rhs
wprim = sp.diff(u, x, 1)
wbis = sp.diff(u, x, 2)

rv1 = sp.Eq(u.subs(x, 0), 0)#w=0 
rv2 = sp.Eq(wprim.subs(x, 0), 0)#wprim=0
rv3 = sp.Eq(u.subs(x, L), 0)#w=0
rv4 = sp.Eq(wbis.subs(x, L), 0)#wbis=0

C1, C2, C3, C4 = sp.symbols("C1, C2, C3, C4")

integrationskonstanter = sp.solve([rv1, rv2, rv3, rv4], [C1, C2, C3, C4])

u= u.subs(integrationskonstanter)
uprim = wprim.subs(integrationskonstanter)
display(u)
rotation = uprim.subs(x, L)
display(rotation)
```
