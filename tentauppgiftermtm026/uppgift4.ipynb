```python
from mtm026 import *
E, delta, v, a, b, r, A, B, u = sp.symbols("E, delta, v, a, b, r, A, B, u")

b = 2*a
u = 1/E*(A*(1-v)*r + (1+v)*(B/r))
RV1 = sp.Eq(u.subs(r, a), delta)
RV2 = sp.Eq(u.subs(r, b), -delta)

sol = sp.solve([RV1, RV2], [A, B])
u = u.subs(sol)
displayvar("u", u)
#spänning
sig_r = A - B/r**2
sig_r = sig_r.subs(sol)

displayvar("sig_r", sig_r)

p_i = -sig_r.subs(r, a)
p_y = -sig_r.subs(r, b)

displayvar("p_i", sp.simplify(p_i))
displayvar("p_y", sp.simplify(p_y))



```
