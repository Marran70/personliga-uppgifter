```python
#först med förskutningsmetod och symmetri
from mtm026 import *
L, E, I, q0 = sp.symbols("L, E, I, q0")

K = sp.zeros(6, 6)
ke = Ke_balk(EI=E*I, L=L)
assem(K, ke, dofs=[1, 2, 3, 4])
assem(K, ke, dofs=[3, 4, 5, 6])
a1, a2, a3, a4, a5, a6 = sp.symbols("a1, a2, a3, a4, a5, a6")
r1, r2, r3, r4, r5, r6 = sp.symbols("r1, r2, r3, r4, r5, r6")
a = sp.Matrix([0, 0, a3, a4, 0, 0])
r = sp.Matrix([r1, r2, 0, 0, r5, r6])
display(K)

f = sp.zeros(6, 1)
fe1 = fe_balk(-q0, L=L) + fe_balk_linjär_last(q0, L=L)
fe2 = fe_balk_linjär_last(-q0, L=L)
assem(f, fe1, dofs=[1, 2, 3, 4])
assem(f, fe2, dofs=[3, 4, 5, 6])
display(f)

ekv_sys = sp.Eq(K*a, f+r)
display(ekv_sys)
obekanta = [a3, a4, r1, r2, r5, r6]

sol = sp.solve(ekv_sys, obekanta)
display(sol)
a = a.subs(sol)
r = r.subs(sol)
display(a, r)
```

```python
L, E, I, q0, q, x = sp.symbols("L, E, I, q0, q, x")
q = -q0*x/L
w = sp.Function("w")(x)

diffekv = sp.Eq(sp.diff(E*I*sp.diff(w, x, 2), x, 2), q)
display(diffekv)
C1, C2, C3, C4 = sp.symbols("C1, C2, C3, C4")
sol = sp.dsolve(diffekv)
display(sol)
u = sol.rhs
wprim= sp.diff(u, x, 1)
wtris=sp.diff(u, x, 3)
display(wprim)
display(wtris)

Rv1 = sp.Eq(u.subs(x, 0), 0)
Rv2 = sp.Eq(wprim.subs(x, 0), 0)
Rv3 = sp.Eq(wprim.subs(x, L), 0)
Rv4 = sp.Eq(wtris.subs(x, L), 0)

integrationskonstanter = sp.solve([Rv1, Rv2, Rv3, Rv4], [C1, C2, C3, C4])
display(integrationskonstanter)

u = u.subs(integrationskonstanter)
display(u)
display(u.subs(x, L))
```
