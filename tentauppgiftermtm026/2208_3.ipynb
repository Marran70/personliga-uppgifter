```python
import sympy as sp
import numpy as np

sigmaz, sigmay, sigmax, sigmae = sp.symbols("sigma_z, sigma_y, sigma_x, sigmae")
p, E, v = sp.symbols("p, E, v")

eq1 = sp.Eq(sigmaz, p)
eq2 = sp.Eq(sigmax, sigmay)
eq3 = sp.Eq(sigmax, v*(sigmay+sigmaz))

lasning = sp.solve([eq1, eq2, eq3], (sigmax, sigmay, sigmaz))

display(lasning)


eq4 = sp.Eq(sigmae, sp.sqrt((sigmax**2+sigmay**2+sigmaz**2 - sigmax*sigmay - sigmay*sigmaz - sigmaz*sigmax)))

vonmises_uttryck = eq4.rhs.subs(lasning)
display(vonmises_uttryck)
förenklat = sp.simplify(vonmises_uttryck)
display(förenklat)


vonmises_kvot = förenklat / eq1.rhs

# 3. Förenkla hela bråket (så att p försvinner och termerna snyggas till)
vonmises_förhållande = sp.simplify(vonmises_kvot)

# 4. Visa det snygga slutresultatet
print("Förhållandet (von Mises / p):")
display(vonmises_förhållande)
```
