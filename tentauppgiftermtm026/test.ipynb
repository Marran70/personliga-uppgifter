```python
from mtm026 import *

sx, sy, sz, txy, tyz, txz, ss = sp.symbols("sx, sy, sz, txy, tyz, txz, ss")

von = sp.Eq(sp.sqrt(sx**2+sy**2+sz**2-sx*sy-sy*sz-sz*sx+3*(txy**2)+3*(tyz**2)+3*(txz**2)), ss)
display(von)

varden = {
    sx: 161e6,
    sy: -53.8e6,
    sz: -107.7e6,
    txy: 107.7e6,
    txz: -80.8e6,
    ss: 450e6
}

sol = sp.solve(von.subs(varden), tyz)
display(sol)
```
