```python
from mtm026 import *
E, I, a0, L, P, = sp.symbols("E, I, a0, L, P")
K = sp.zeros(9, 9)
EI =E*I
k = a0*EI/L**3
```

```python
ke1 = Ke_balk(EI=EI, L=L/2)
ke3 = Ke_balk(EI=EI, L=L/2)
ke2 = Ke_balk(EI=2*EI, L=L)
fe = Ke_fjäder(k=k)
assem(K, ke1, dofs=[1, 2, 3, 4])
assem(K, ke3, dofs=[3, 4, 5, 6])
assem(K, ke2, dofs=[5, 6, 7, 8])
assem(K, fe, dofs=[9, 7])
```

```python
a1, a2, a3, a4, a5, a6, a7, a8, a9 = sp.symbols("a1:10")
r1, r2, r3, r4, r5, r6, r7, r8, r9 = sp.symbols("r1:10")

a = sp.Matrix([0, 0, a3, a4, 0, a6, a7, a8, 0])
f = sp.Matrix([r1, r2, -P, 0, r5, 0, 0, 0, r9])

ekv_sys = sp.Eq(K*a, f)
display(ekv_sys)
obekanta = [a3, a4, a6, a7, a8, r1, r2, r5, r9]
sol = sp.solve(ekv_sys, obekanta)
display(a.subs(sol))
display(f.subs(sol))
```
