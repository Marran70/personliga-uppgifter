```python
from mtm026 import *
b, t, A, p, g, E, I, L, q0, P, v = sp.symbols("b, t, A, p, g, E, I, L, q0, P, v")
E = 200e9

p = 0
b = 0.01
t = 0.002
L = 1
P = 100
g = 9.81
v = 0.3
I= t*b**3/6 + t*b*b**2/2
A = b**2 -(b-t)**2
```

```python
K = sp.zeros(6, 6)

Ke = Ke_balk(EI = E*I, L = L)
assem(K, Ke, dofs=[1, 2, 3, 4])
assem(K, Ke, dofs=[3, 4, 5, 6])
```

```python
a1, a2, a3, a4, a5, a6 = sp.symbols("a1:7")
f1, f2, f3, f4, f5, f6 = sp.symbols("r1:7")

d = sp.Matrix([0, 0, a3, a4, 0, 0])
r = sp.Matrix([f1, f2, 0, 0, f5, f6])

f = sp.zeros(6, 1)
fe = fe_balk(-p*g*A, L)
assem(f, fe, dofs=[1, 2, 3, 4])
assem(f, fe, dofs=[3, 4, 5, 6])
f[2] += -P
```

```python
ekv_sys = sp.Eq(K*d, f+r)
display(ekv_sys)
```

```python
obekanta = [a3, a4, f1, f2, f5, f6]
sol = sp.solve(ekv_sys, obekanta)
displayvar("a3", sol[a3])
displayvar("a4", sol[a4])
```
