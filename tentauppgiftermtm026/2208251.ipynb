```python
from mtm026 import *
E, I, L, q, q0, x, Q, w = sp.symbols("E, I, L, q, q0, x, Q, w")
EI = E*I
```

```python

q = -q0/L**2*(x*L-x**2)
```

```python
w = sp.Function('w')(x)
ode = sp.Eq(sp.diff(EI*sp.diff(w, x, 2), x, 2), q)
display(ode)
```

```python
sol = sp.dsolve(ode)
display(sol)
w = sol.rhs
display(sol)
```

```python
#randvillkor
wbis = w.diff(x, 2)

M = EI*wbis

r1 = sp.Eq(w.subs(x, 0), 0)
r2 = sp.Eq(wbis.subs(x, 0), 0)
r3 = sp.Eq(w.subs(x, L), 0)
r4 = sp.Eq(wbis.subs(x, L), 0)

display(r1, r2, r3, r4)
```

```python
ekvationer = [r1, r2, r3, r4]
obekanta = sp.symbols("C1, C2, C3, C4")
konstanter = sp.solve(ekvationer, obekanta)
display(konstanter)
w = w.subs(konstanter)
display(w)

M = EI*wbis.subs(konstanter)
display(M)
```
