```python
from mtm026 import *

L, q, P, E, A, I = sp.symbols("L, q, P, E, A, I")
EI1 = E*I
EA = E*I/(L**2)
q = 2*P/L
```

```python
Dofs = [
    [1, 2, 3, 4],
    [3, 4, 5, 6],
    [7, 3]
]
a1, a2, a3, a4, a5, a6,a7 = sp.symbols("a1, a2, a3, a4, a5, a6, a7")
r1, r2, r7 = sp.symbols("r1, r2, r7")

K = sp.zeros(7, 7)
ke1 = Ke_balk(EI=EI1, L=L)
assem(K, ke1, dofs=Dofs[0])
assem(K, ke1, dofs=Dofs[1])

ke2 = Ke_fjäder(EA/L)
assem(K, ke2, dofs=Dofs[2])
display(K)

```

```python
frihetsgrader = sp.Matrix([0, 0, a3, a4, a5, a6, 0])
reaktionskrafter = sp.Matrix([r1, r2, 0, 0, 0, 0, r7])
f = sp.zeros(7, 1)
fe = fe_balk(q=-q, L=L)
assem(f, fe, dofs=Dofs[0])
f[4] += -P
display(f)
```

```python
ekv_sys = sp.Eq(K*frihetsgrader, f+reaktionskrafter)
display(ekv_sys)
obekanta = [a3, a4, a5, a6, r1, r2, r7]
sol = sp.solve(ekv_sys, obekanta)
displayvar("a_5", sol[a5])
displayvar("r2", sol[r2])

display_eqnsys(K, frihetsgrader, f+reaktionskrafter)
```
