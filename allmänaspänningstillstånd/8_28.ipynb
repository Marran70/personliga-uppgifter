```python
from mtm026 import *

S = np.array([
    [80, 40, 0],
    [40, 0, 20],
    [0, 20, 80]
], dtype=float)

n = (1 / np.sqrt(6)) * np.array([[1], [2], [1]])

s = S @ n

nSp = n.T @ s

tau = np.sqrt(np.linalg.norm(s)**2 - nSp**2)

v,W = eig(S)

huvudsp = v
ordn = np.argsort(huvudsp)[::-1]
sorteradeHuvudsp = huvudsp[ordn]

sorteradeHuvudspRiktningar = W[:, ordn]

displayvar("s", s)
displayvar("σ_n", nSp)
displayvar("τ", tau)
displayvar("σ", sorteradeHuvudsp)
displayvar("n", sorteradeHuvudspRiktningar)
```
